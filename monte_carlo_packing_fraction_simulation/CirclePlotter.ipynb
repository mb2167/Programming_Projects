```python
# This program plots the circle coordinates obtained from the packing fraction

import matplotlib.pyplot as plt
import numpy as np
import matplotlib

# Set font to Times New Roman
matplotlib.rcParams['font.family'] = 'Times New Roman'

def read_coordinates(filename):
    with open(filename, 'r') as file:
        return [tuple(map(float, line.split())) for line in file]  # Split by spaces

def plot_circles(coordinates, radius):
    # Increase the figsize and dpi for higher resolution
    fig, ax = plt.subplots(figsize=(20, 20), dpi=100)

    # Plot the circles
    for x, y in coordinates:
        circle = plt.Circle((x, y), radius, edgecolor='blue', facecolor='none')
        ax.add_patch(circle)

    # Set the limits of the axes
    ax.set_xlim(0, 12.5)
    ax.set_ylim(0, 12.5)

    # Add vertical lines to show MPI border when MPI tasks = 10
    plt.axvline(x=20, color='green', linestyle='--', linewidth=2)
    
    # Increase the font size of the axis numbers
    ax.tick_params(axis='both', which='major', labelsize=60)

    plt.show()

coordinates = read_coordinates('coordinates.txt')
plot_circles(coordinates, 1.234)
```
